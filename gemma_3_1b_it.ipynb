````python
# 27-Question Multilingual Medical Knowledge Test

```python
# ============================================================
# Gemma Model - 27 Question Multilingual Medical Knowledge Test
# Domains:
# 1. Ayurveda - 9 questions
# 2. Traditional Chinese Medicine (TCM) - 9 questions
# 3. Allopathy / Modern Medicine - 9 questions
#
# Languages in each domain:
# English - 3
# Hindi   - 3
# Tamil   - 3
# ============================================================

questions = [

    # ========================================================
    # AYURVEDA - ENGLISH (3)
    # ========================================================

    # "What are the three doshas in Ayurveda?",

    # "Which dosha is primarily responsible for movement and bodily motion according to Ayurveda?",

    # "What are the five elements associated with Vata, Pitta, and Kapha?",


    # ========================================================
    # AYURVEDA - HINDI (3)
    # ========================================================

    "आयुर्वेद में तीन दोष कौन से हैं?",

    "आयुर्वेद के अनुसार शरीर में गति और संचलन के लिए कौन सा दोष जिम्मेदार है?",

    "आयुर्वेद में वात, पित्त और कफ किन पाँच तत्वों से जुड़े हैं?",


    # ========================================================
    # AYURVEDA - TAMIL (3)
    # ========================================================

    "ஆயுர்வேதத்தில் மூன்று தோஷங்கள் எவை?",

    "ஆயுர்வேதத்தின் படி உடலில் இயக்கம் மற்றும் அசைவுக்கு பொறுப்பான தோஷம் எது?",

    "ஆயுர்வேதத்தில் வாதம், பித்தம் மற்றும் கபம் ஆகியவை எந்த ஐந்து மூலக்கூறுகளுடன் தொடர்புடையவை?",


    # ========================================================
    # TRADITIONAL CHINESE MEDICINE (TCM) - ENGLISH (3)
    # ========================================================

    "What do Yin and Yang represent in Traditional Chinese Medicine?",

    "What is Qi in Traditional Chinese Medicine?",

    "What are the Five Phases in Traditional Chinese Medicine?",


    # ========================================================
    # TCM - HINDI (3)
    # ========================================================

    "पारंपरिक चीनी चिकित्सा में यिन और यांग का क्या अर्थ है?",

    "पारंपरिक चीनी चिकित्सा में Qi का क्या अर्थ है?",

    "पारंपरिक चीनी चिकित्सा में Five Phases या पाँच चरण क्या हैं?",


    # ========================================================
    # TCM - TAMIL (3)
    # ========================================================

    "பாரம்பரிய சீன மருத்துவத்தில் யின் மற்றும் யாங் எதைக் குறிக்கின்றன?",

    "பாரம்பரிய சீன மருத்துவத்தில் Qi என்றால் என்ன?",

    "பாரம்பரிய சீன மருத்துவத்தில் Five Phases அல்லது ஐந்து நிலைகள் என்ன?",


    # ========================================================
    # ALLOPATHY / MODERN MEDICINE - ENGLISH (3)
    # ========================================================

    "What is the normal resting heart rate for most healthy adults?",

    "What is the main function of red blood cells?",

    "What is insulin and how does it regulate blood sugar?",


    # ========================================================
    # ALLOPATHY / MODERN MEDICINE - HINDI (3)
    # ========================================================

    "अधिकांश स्वस्थ वयस्कों की सामान्य विश्राम अवस्था में हृदय गति कितनी होती है?",

    "लाल रक्त कोशिकाओं का मुख्य कार्य क्या है?",

    "इंसुलिन क्या है और यह रक्त शर्करा के स्तर को कैसे नियंत्रित करता है?",


    # ========================================================
    # ALLOPATHY / MODERN MEDICINE - TAMIL (3)
    # ========================================================

    "பெரும்பாலான ஆரோக்கியமான பெரியவர்களின் இயல்பான ஓய்வு நேர இதயத் துடிப்பு எவ்வளவு?",

    "சிவப்பு இரத்த அணுக்களின் முக்கிய செயல்பாடு என்ன?",

    "இன்சுலின் என்றால் என்ன? அது இரத்தத்தில் உள்ள சர்க்கரையின் அளவை எவ்வாறு கட்டுப்படுத்துகிறது?"
]


# ============================================================
# VERIFY QUESTION COUNT
# ============================================================

print("Total Questions:", len(questions))

assert len(questions) == 27

print("Question validation successful!")
```

---

# Load the Gemma Model

```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

MODEL_NAME = "google/gemma-3-1b-it"

print("Loading model:", MODEL_NAME)


# ------------------------------------------------------------
# 1. LOAD TOKENIZER
# ------------------------------------------------------------

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token


# ------------------------------------------------------------
# 2. LOAD MODEL
# ------------------------------------------------------------

model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    dtype=torch.float16 if torch.cuda.is_available() else torch.float32,
    device_map="auto"
)

model.eval()

print("Model loaded successfully!")
```

---

# Generate Answer Function

```python
def generate_answer(question):

    messages = [
        {
            "role": "user",
            "content": question
        }
    ]

    prompt = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        prompt,
        return_tensors="pt"
    )

    inputs = {
        key: value.to(model.device)
        for key, value in inputs.items()
    }

    with torch.no_grad():

        outputs = model.generate(
            **inputs,
            max_new_tokens=150,
            do_sample=False,
            pad_token_id=tokenizer.eos_token_id
        )

    input_length = inputs["input_ids"].shape[1]

    generated_tokens = outputs[0][input_length:]

    answer = tokenizer.decode(
        generated_tokens,
        skip_special_tokens=True
    ).strip()

    return answer
```

---

# Run All 27 Questions

```python
# ============================================================
# RUN ALL 27 QUESTIONS
# ============================================================

results = []

for i, question in enumerate(questions, start=1):

    print("\n" + "=" * 80)
    print(f"QUESTION {i}/27")
    print("=" * 80)

    print("\nQuestion:")
    print(question)

    answer = generate_answer(question)

    print("\nModel Output:")
    print(answer)

    results.append({
        "id": i,
        "question": question,
        "output": answer
    })


print("\n")
print("=" * 80)
print("ALL 27 QUESTIONS COMPLETED")
print("=" * 80)
```

# Test distribution

| Domain                      | English | Hindi | Tamil |  Total |
| --------------------------- | ------: | ----: | ----: | -----: |
| Ayurveda                    |       3 |     3 |     3 |      9 |
| TCM                         |       3 |     3 |     3 |      9 |
| Allopathy / Modern Medicine |       3 |     3 |     3 |      9 |
| **Total**                   |   **9** | **9** | **9** | **27** |
````
